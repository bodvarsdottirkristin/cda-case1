```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder

# column groups
y_col = "y"
num_cols = [f"x_{i:02d}" for i in range(1, 96)]
cat_cols = [f"C_{i:02d}" for i in range(1, 6)]

# separate target
y = df[y_col]
X = df.drop(columns=[y_col])

# ----- Encode categorical variables -----
encoder = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)
X_cat = encoder.fit_transform(X[cat_cols])

# ----- Scale numerical variables -----
scaler = StandardScaler()
X_num = scaler.fit_transform(X[num_cols])

# ----- Combine -----
import numpy as np
X_combined = np.hstack([X_num, X_cat])

# ----- KNN Imputation -----
imputer = KNNImputer(n_neighbors=5, weights="distance")
X_imputed = imputer.fit_transform(X_combined)

# ----- Split back into numeric and categorical -----
X_num_imputed = X_imputed[:, :len(num_cols)]
X_cat_imputed = X_imputed[:, len(num_cols):]

# inverse scaling of numeric variables
X_num_imputed = scaler.inverse_transform(X_num_imputed)

# convert back to dataframe
df_num = pd.DataFrame(X_num_imputed, columns=num_cols, index=df.index)
df_cat = pd.DataFrame(X_cat_imputed, columns=cat_cols, index=df.index)

# round categorical values to nearest category
df_cat = df_cat.round().astype(int)

# reconstruct dataframe
df_imputed = pd.concat([y, df_num, df_cat], axis=1)
```

## Random Forest A

```python
DATA_DIR2 = Path('../data/raw')
df_Xnew  = pd.read_csv(DATA_DIR2 / 'case1Data_Xnew.csv')
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# ── 1. Define feature columns ────────────────────────────────────────────────
categorical_cols = ['C_01', 'C_03', 'C_04', 'C_05']
numerical_cols   = [f'x_{i:02d}' for i in range(1, 96)]

X = df_imputed[numerical_cols + categorical_cols]
y = df_imputed['y']
```

```python
# ── 2. Build preprocessing + model pipeline ──────────────────────────────────
preprocessor = ColumnTransformer(transformers=[
    ('num', 'passthrough', numerical_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_cols)
])

pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
# ── 3. Nested cross-validation (10 outer / 5 inner folds) ────────────────────
from sklearn.model_selection import GridSearchCV, KFold

# Hyperparameter grid for the inner loop to search over
param_grid = {
    'model__n_estimators':    [100, 300],
    'model__max_features':    ['sqrt', 0.33],
    'model__min_samples_leaf': [3, 5, 10],
}

outer_cv = KFold(n_splits=10, shuffle=True, random_state=42)
inner_cv = KFold(n_splits=5,  shuffle=True, random_state=42)

outer_rmse_scores = []
best_params_per_fold = []

for fold, (train_idx, test_idx) in enumerate(outer_cv.split(X), start=1):
    X_train_outer, X_test_outer = X.iloc[train_idx], X.iloc[test_idx]
    y_train_outer, y_test_outer = y.iloc[train_idx], y.iloc[test_idx]

    # Inner loop: hyperparameter search on the training split
    grid_search = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grid,
        cv=inner_cv,
        scoring='neg_root_mean_squared_error',
        refit=True,       # refit best model on the full outer-train split
        n_jobs=-1
    )
    grid_search.fit(X_train_outer, y_train_outer)

    # Outer evaluation: predict on the held-out test split
    best_model = grid_search.best_estimator_
    preds = best_model.predict(X_test_outer)
    rmse  = np.sqrt(mean_squared_error(y_test_outer, preds))
    outer_rmse_scores.append(rmse)
    best_params_per_fold.append(grid_search.best_params_)

    print(f"  Fold {fold:02d} | RMSE: {rmse:.4f} | Best params: {grid_search.best_params_}")

print(f"\n  Average RMSE across {len(outer_rmse_scores)} outer folds: {np.mean(outer_rmse_scores):.4f}")
print(f"  Std Dev of RMSE:                                    {np.std(outer_rmse_scores):.4f}")
```

```python
# ── 4. Retrain final model on the full dataset ───────────────────────────────
# Run one final inner-CV grid search over ALL data to pick the best params,
# then the refitted model is used for prediction.
final_grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=inner_cv,
    scoring='neg_root_mean_squared_error',
    refit=True,
    n_jobs=-1
)
final_grid_search.fit(X, y)
print(f"\n  Final model best params: {final_grid_search.best_params_}")
```

```python
# ── 5. Prepare df_Xnew and predict ──────────────────────────────────────────
df_Xnew_clean = df_Xnew.drop(columns=['C_02'], errors='ignore')
X_new = df_Xnew_clean[numerical_cols + categorical_cols]

y_pred_new = final_grid_search.best_estimator_.predict(X_new)
```

```python
# Plot percentage frequency of predicted values and actual values
plt.figure(figsize=(10, 6))
sns.histplot(y_pred_new, bins=30, color='blue', label='Predicted', kde=True, stat='density')
sns.histplot(y, bins=30, color='orange', label='Actual', kde=True, stat='density', alpha=0.6)
plt.title('Distribution of Predicted vs Actual Values')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.show()
```
